```python
import os
```

```python
url = "YOUTUBE URL"
video = YouTube(url)
```

```python
print('Title:',video.title)
```

```python
out_path = video.streams.filter(only_audio=True).first().download()
```

```python
os.rename(out_path, "output.mp3")
```
